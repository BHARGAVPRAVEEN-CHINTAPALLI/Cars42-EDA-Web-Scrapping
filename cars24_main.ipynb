```python
def scrape_cars24_lists(file_path):

    file_name = os.path.basename(file_path)
    location_value = os.path.splitext(file_name)[0]

    name = []
    price = []
    emi_per_month = []
    driven_kms = []
    engine_type = []
    transmission = []
    number_type = []
    location = []
    status = []
    variant = []

    with open(file_path, "r", encoding="utf-8") as f:
        soup = BeautifulSoup(f, "html.parser")

    cards = soup.find_all("div", {"class": "styles_normalCardWrapper__qDZjq"})

    for card in cards:

        title_tag = card.find("span", {"class": "sc-bcXHqh bAcffq"})
        if not title_tag:
            continue

        name.append(title_tag.text.strip())
        location.append(location_value)

        price_tag = card.find("p", {"class": "sc-bcXHqh hvRpEM"})
        price_value = price_tag.text.strip() if price_tag else None
        price.append(price_value)

        variant_tag = card.find("span", {"class": "sc-bcXHqh bKVBht"})
        variant.append(variant_tag.text.strip() if variant_tag else None)

        emi_tag = card.find(string=lambda text: "EMI" in text if text else False)
        emi_per_month.append(emi_tag.strip() if emi_tag else None)

        booked_tag = card.find(string=lambda text: "BOOKED" in text if text else False)
        status.append("BOOKED" if booked_tag else "Available")

        detail_tags = card.find_all("p", {"class": "sc-bcXHqh kNDBvu"})
        values = [tag.text.strip() for tag in detail_tags if tag.text.strip()]

        km = next((v for v in values if "km" in v.lower()), None)
        fuel = next((v for v in values if v in ["Petrol", "Diesel", "CNG", "Electric"]), None)
        trans = next((v for v in values if v in ["Manual", "Auto"]), None)
        reg = next((v for v in values if re.match(r"[A-Z]{2}-\d+", v)), None)

        driven_kms.append(km)
        engine_type.append(fuel)
        transmission.append(trans)
        number_type.append(reg)

    return (name, price, emi_per_month, driven_kms,
            engine_type, transmission, number_type,
            location, status, variant)
```

## Using Regular Expressions cleaning the unwanted characters from a specific columns

```python
# Seperating the Year, Brand and Model from Name Column
pattern = r'(?P<Year>\d{4})\s+(?P<Brand>\w+)\s+(?P<Model>.+)'
data[['Year', 'Brand', 'Model']] = data['Name'].str.extract(pattern)

# Removing unwanted Symbols from Emi_Per_Month column
pattern = '(\\d+(?:,\\d+)*)'
data['EMI_per_Month'] = data['EMI_per_Month'].str.extract(pattern)
data['EMI_per_Month'] = data['EMI_per_Month'].str.replace(',', '')

# Extracting the numerical part from Price column
pattern = r'([\d.]+)'
data['Price(In_Lakhs)'] = data['Price(In_Lakhs)'].str.extract(pattern)

# Extracting the numerical part from the Driven_Kms column
pattern = r'\D'
data['Driven_Kms'] = data['Driven_Kms'].str.replace(pattern, '', regex=True)
```

## Univariate Analysis

```python
plt.figure(figsize=(15,5))
sns.kdeplot(data=data,x=data["Price(In_Lakhs)"],label="Price In Lakhs")
plt.title("Density plot of Prices all over India")
plt.axvline(x=data["Price(In_Lakhs)"].mean(),color="red",label=f"mean:{data["Price(In_Lakhs)"].mean():.2f}",linestyle=":")
plt.axvline(x=data["Price(In_Lakhs)"].median(),color="green",label=f"median{data["Price(In_Lakhs)"].median():.2f}",linestyle="--")
plt.legend()
plt.show()
```

## Observations from the Plot
### - Right-skewed distribution:
#### Most cars are priced at the lower end (around 3–6 Lakhs), with fewer cars at higher prices. This long tail indicates that premium cars exist but are rare compared to budget/mid-range cars.
## - Mean vs. Median difference:
### - Mean ≈ 5.40 Lakhs
### - Median ≈ 4.31 Lakhs
#### The mean is higher than the median, which confirms the skewness — a few very expensive cars are pulling the average upward.
## - Concentration zone:
### The density curve peaks around 3–5 Lakhs, suggesting this is the most common resale price range in India. It reflects affordability and demand in the used car market.
## - Market segmentation hint:
### The distribution suggests three broad segments:
#### - Budget cars: 2–5 Lakhs (majority of listings)
#### - Mid-range cars: 6–10 Lakhs (smaller but significant share)
#### - Premium cars: >10 Lakhs (rare, niche market)
